```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Style (important for professional look)
sns.set_theme(style="whitegrid", context="talk")
```

```python
data = pd.DataFrame({
    "Region": ["Denmark", "Global"],
    "Chargers_per_million": [8333, 662]
})

plt.figure(figsize=(8,5))
ax = sns.barplot(data=data, x="Region", y="Chargers_per_million")

# Labels
ax.set_title("Public EV Chargers per Million People", fontsize=16, weight="bold")
ax.set_ylabel("Chargers per million inhabitants")
ax.set_xlabel("")

# Annotate values
for i, v in enumerate(data["Chargers_per_million"]):
    ax.text(i, v + 200, f"{v:,}", ha='center', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
data = pd.DataFrame({
    "Region": ["Denmark", "Global"],
    "EVs_per_charger": [11, 10]
})

plt.figure(figsize=(8,5))
ax = sns.barplot(data=data, x="Region", y="EVs_per_charger")

ax.set_title("EVs per Public Charger (System Load)", fontsize=16, weight="bold")
ax.set_ylabel("EVs per charger")
ax.set_xlabel("")

for i, v in enumerate(data["EVs_per_charger"]):
    ax.text(i, v + 0.2, f"{v}", ha='center', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
data = pd.DataFrame({
    "Region": ["Denmark", "Global"],
    "Daily_charging_events": [50000, 3000000]
})

plt.figure(figsize=(8,5))
ax = sns.barplot(data=data, x="Region", y="Daily_charging_events")

ax.set_title("Estimated Daily Public Charging Demand", fontsize=16, weight="bold")
ax.set_ylabel("Charging events per day")
ax.set_xlabel("")

# Format large numbers
for i, v in enumerate(data["Daily_charging_events"]):
    ax.text(i, v * 1.02, f"{v/1000:.0f}k" if v < 1e6 else f"{v/1e6:.1f}M", 
            ha='center', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
import numpy as np

time = np.linspace(0, 24, 100)

# Normal demand
baseline = 0.6 + 0.2*np.sin((time-8)/24*2*np.pi)

# Peak demand
peak = baseline + 0.25*np.exp(-((time-17)**2)/10)

# Disruption scenario
disruption = peak * 1.4

capacity = np.ones_like(time)

plt.figure(figsize=(10,6))

plt.plot(time, baseline, label="Normal demand", linewidth=2)
plt.plot(time, peak, label="Peak demand", linewidth=2)
plt.plot(time, disruption, label="Disruption (cold / congestion)", linewidth=2)

plt.axhline(1.0, linestyle="--", label="Capacity limit")

plt.title("System Load vs Capacity (Resilience Perspective)", fontsize=16, weight="bold")
plt.xlabel("Time of day")
plt.ylabel("Load (fraction of capacity)")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
stations = ["Hub", "Service Area", "Exit", "Rural"]
utilization = [0.9, 0.6, 0.3, 0.1]

data = pd.DataFrame({
    "Station_type": stations,
    "Utilization": utilization
})

plt.figure(figsize=(8,5))
ax = sns.barplot(data=data, x="Station_type", y="Utilization")

ax.set_title("Utilization by Station Type", fontsize=16, weight="bold")
ax.set_ylabel("Average utilization")
ax.set_xlabel("")

for i, v in enumerate(utilization):
    ax.text(i, v + 0.03, f"{v:.1f}", ha='center', fontsize=12)

plt.tight_layout()
plt.show()
```
